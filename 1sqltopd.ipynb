```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("mydb.db")
cursor = conn.cursor()

with open("world.sql", "r", encoding="latin-1") as f:
    cursor.executescript(f.read())

tables = cursor.execute("SELECT name FROM sqlite_master WHERE type='table';").fetchall()
print("Tables:", tables)

df = pd.read_sql(f"SELECT * FROM {tables[0][0]}", conn)
print(df.head())
```
