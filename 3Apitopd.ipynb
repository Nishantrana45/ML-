```python
import requests
import pandas as pd
import time

df_list = []

for i in range(1, 501):
    try:
        response = requests.get(
            f"https://api.themoviedb.org/3/movie/top_rated?api_key=17ef6b517c403d7a73bf6f3d7ebcd200&language=en-US&page={i}"
        )
        
        data = response.json()
        
        temp_df = pd.DataFrame(data["results"])[
            ["id", "title", "overview", "release_date", "popularity"]
        ]
        df_list.append(temp_df)

        print(f"Page {i} done")

        time.sleep(0.3)   # 🔥 IMPORTANT (300ms delay)

    except Exception as e:
        print(f"Error on page {i}: {e}")
        continue

df = pd.concat(df_list, ignore_index=True)
```
